## 7. Frequently Asked Questions (FAQ Answers)

### Q1: What is prompt engineering, and why is it important for LLM-based applications?
**Answer:** Prompt engineering is the art and science of formulating clear, structured input instructions (prompts) to guide a Large Language Model toward producing accurate, relevant, contextually sound, and correctly formatted outputs. It plays a critical role in controlling model persona, constraining boundaries, eliminating hallucinations, and maximizing model performance.

### Q2: Differentiate between zero-shot, few-shot, and chain-of-thought prompting.
| Prompting Technique | Description | Ideal Use Case |
| :--- | :--- | :--- |
| **Zero-Shot Prompting** | Provides instructions without any prior input-output examples. | Standard QA, factual retrieval, sentiment classification. |
| **Few-Shot Prompting** | Includes 2–5 exemplar input-output pairs within the prompt context. | Complex output formatting, domain-specific terminology translation. |
| **Chain-of-Thought (CoT)** | Instructs the model to reason step-by-step ("Let's think step-by-step"). | Mathematical reasoning, multi-step logic, complex problem solving. |

### Q3: What is a system prompt, and how does it influence chatbot behavior?
**Answer:** A system prompt (or role prompt) sets the fundamental persona, knowledge boundary constraints, tone, and operational rules for the LLM. It acts as an immutable guiding policy across all conversation turns (e.g., forcing a healthcare bot to include disclaimers or restricting a college bot to campus-specific facts).

### Q4: How is context/chat history managed in a multi-turn conversational AI system?
**Answer:** Context management retains the thread of conversation across multiple user turns. This is achieved by maintaining an ordered list of message objects `[{"role": "system", ...}, {"role": "user", ...}, {"role": "assistant", ...}]` bound to a unique `session_id`. On each new turn, the entire message history is appended and sent to the LLM API.

### Q5: What are intents and dialogue states in a chatbot workflow?
**Answer:**
- **Intent**: The specific goal or purpose behind a user's input (e.g., `admissions_inquiry`, `fee_structure`, `emergency_escalation`).
- **Dialogue State**: The current node or step in a conversational state machine (e.g., `GREETING`, `AWAITING_CRITERIA`, `BOOKING_CONFIRMATION`, `ESCALATION`).

### Q6: What safety measures should be built into a healthcare or sensitive-domain chatbot?
**Answer:**
1. **Mandatory Safety Disclaimers**: Informing users that the chatbot is an AI assistant, not a licensed physician.
2. **Emergency Escalation Triggers**: Hard-coded safety hooks that immediately redirect severe symptoms (e.g., chest pain, respiratory distress) to emergency hotline numbers (108/911).
3. **No Prescriptions or Diagnosis**: Restricting output to general wellness guidance without prescribing drug dosages.
