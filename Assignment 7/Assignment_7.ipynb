## 3. Case Study 1: AI Document Summarizer using HuggingFace Pipeline

```
# Initialize Summarization Pipeline
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

document = """
The Transformer architecture was introduced in the landmark 2017 paper "Attention Is All You Need" by Vaswani et al.
Unlike previous sequential neural network architectures such as Recurrent Neural Networks (RNNs) and Long Short-Term Memory networks (LSTMs),
the Transformer relies entirely on attention mechanisms to compute representations of its input and output without using sequence-aligned RNNs or convolution.
This allows for significantly higher parallelization during training and enables models to efficiently capture long-range contextual dependencies across long text sequences.
Transformers serve as the core foundational architecture for state-of-the-art Generative AI models including BERT, BART, T5, GPT-4, and LLaMA.
"""

summary = summarizer(document, max_length=100, min_length=30, do_sample=False)
print("=" * 60)
print("ORIGINAL DOCUMENT WORD COUNT:", len(document.split()))
print("GENERATED ABSTRACTIVE SUMMARY:")
print("=" * 60)
print(summary[0]['summary_text'])
print("SUMMARY WORD COUNT:", len(summary[0]['summary_text'].split()))
```

## 4. Case Study 2: Resume Screening & Extractive Question Answering

```
# Initialize Question-Answering Pipeline
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

resume_context = """
Candidate Profile: Aarav Sharma
Title: Senior Software Engineer
Location: Pune, Maharashtra
Total Experience: 6 years of software engineering experience in Python, React, Flask, FastAPI, PostgreSQL, Docker, and AWS.
Education: Bachelor of Technology (B.Tech) in Information Technology from Sanjivani College of Engineering, Kopargaon with CGPA 8.9/10.
Certifications: AWS Certified Solutions Architect.
Key Achievements: Winner of National Level Hackathon 2020. Reduced API response latency by 35% through Redis caching and PostgreSQL query optimization.
"""

questions = [
    "What is the candidate's name?",
    "How many years of experience does the candidate have?",
    "What programming languages and frameworks does the candidate know?",
    "Which college did the candidate graduate from?",
    "What certification does the candidate hold?"
]

print("=" * 60)
print("CANDIDATE RESUME QUESTION ANSWERING RESULTS:")
print("=" * 60)

for q in questions:
    result = qa_pipeline(question=q, context=resume_context)
    print(f"❓ Question: {q}")
    print(f"👉 Answer  : {result['answer']} (Confidence Score: {result['score']:.4f})")
    print("-" * 50)
```

## 5. Direct Model & Tokenizer API (`AutoTokenizer` & `AutoModelForSeq2SeqLM`)

```
# Load model and tokenizer directly
model_name = "t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

input_text = "summarize: " + document
inputs = tokenizer.encode(input_text, return_tensors="pt", max_length=512, truncation=True)
summary_ids = model.generate(inputs, max_length=100, min_length=25, num_beams=4, early_stopping=True)

decoded_summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("=" * 60)
print("DIRECT T5-SMALL MODEL DECODED SUMMARY:")
print("=" * 60)
print(decoded_summary)
```

## 6. Frequently Asked Questions (FAQ Answers)

### Q1: What is the Transformer architecture, and how does it differ from RNNs/LSTMs?
**Answer:** The Transformer architecture is a neural network model introduced in "Attention Is All You Need" (Vaswani et al., 2017) that relies entirely on self-attention mechanisms without recurrent or convolutional layers. Unlike RNNs/LSTMs that process sequence tokens step-by-step sequentially (causing memory bottlenecks and vanishing gradients), Transformers process all sequence tokens simultaneously in parallel, dramatically accelerating training speed and modeling long-range contextual relationships.

### Q2: Explain the concept of self-attention and multi-head attention.
**Answer:** 
- **Self-Attention**: Computes contextual attention weights between every pair of tokens in a sequence by transforming token embeddings into Query ($Q$), Key ($K$), and Value ($V$) projections: $	ext{Attention}(Q,K,V) = 	ext{Softmax}\left(rac{QK^T}{\sqrt{d_k}}ight)V$.
- **Multi-Head Attention**: Runs multiple self-attention operations in parallel across separate feature subspaces, enabling the model to focus on diverse syntactic, semantic, and positional relationships concurrently.

### Q3: What is the difference between encoder-only, decoder-only, and encoder-decoder models?
| Architecture Type | Examples | Attention Type | Primary Use Case |
| :--- | :--- | :--- | :--- |
| **Encoder-Only** | BERT, RoBERTa, DistilBERT | Bidirectional Self-Attention | Classification, Named Entity Recognition, Extractive QA |
| **Decoder-Only** | GPT-3.5, GPT-4, LLaMA | Autoregressive Causal Attention | Text Generation, Code Generation, Dialogue Systems |
| **Encoder-Decoder** | BART, T5 | Bidirectional + Autoregressive | Translation, Abstractive Summarization, Sequence Transformation |

### Q4: What is the HuggingFace `pipeline()` API, and what tasks does it support?
**Answer:** The `pipeline()` API is an abstracted, high-level wrapper in HuggingFace Transformers that connects tokenization, model inference, and output post-processing into a single unified function. It supports tasks such as `summarization`, `question-answering`, `text-classification`, `translation`, and `zero-shot-classification`.

### Q5: What is the difference between extractive and abstractive text summarization?
**Answer:** 
- **Extractive Summarization**: Identifies and extracts the exact key sentences directly from the source text based on sentence relevance metrics.
- **Abstractive Summarization**: Generates new sentences that capture and rephrase the core information using seq2seq encoder-decoder models (BART/T5), closely mimicking human summary creation.

### Q6: How does an extractive question-answering model predict the answer span from a context?
**Answer:** Extractive QA models (like BERT fine-tuned on SQuAD) pass the question and context together into the encoder. The model outputs logits for each context token representing the probability of that token being the **start position** and **end position** of the answer span within the context text.

### Q7: What is fine-tuning, and why is it useful for domain-specific NLP applications like resume screening?
**Answer:** Fine-tuning is the process of taking a general pretrained transformer model (trained on massive web corpora) and updating its parameters on a smaller, domain-specific dataset (such as technical resumes or legal documents). Fine-tuning adapts the model to domain terminology, jargon, and specialized syntax while leveraging general language understanding.
